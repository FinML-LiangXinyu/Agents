```python
# 流式输出 agent 响应
reponse = requests.post(
    "http://127.0.0.1:8000/chat/stream/Anti-Fraud-Agent",
    json = {"user_name": "user", "user_input": f"这是诈骗短信吗？{isFraud}"}
)
reponse.text
```

```python
# 结构化输出 agent 响应
reponse = requests.post(
    "http://127.0.0.1:8000/chat/generate/Anti-Fraud-Agent",
    json = {"user_name": "user", "user_input": f"这是诈骗短信吗？【{isFraud}】"}
)
json.loads(reponse.json())
```
